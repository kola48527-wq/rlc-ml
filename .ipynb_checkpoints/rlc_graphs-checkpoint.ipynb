```python
import numpy as np
import matplotlib
matplotlib.use('Agg') 
import matplotlib.pyplot as plt
import itertools
import os
import random

output_dir = "rlc_graphs"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
R_list = [1, 2, 4, 8, 10, 15, 20, 25, 30, 40, 50, 60, 75, 100, 120, 150, 200, 300, 400, 500]
L_list = [0.005, 0.01, 0.02, 0.03, 0.05, 0.07, 0.1, 0.12, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0]
C_list = [1e-7, 2e-7, 5e-7, 1e-6, 2e-6, 3e-6, 4e-6, 5e-6, 8e-6, 1e-5, 1.5e-5, 2e-5, 3e-5, 4e-5, 5e-5, 6e-5, 7e-5, 8e-5, 9e-5, 1e-4]
U_list = [5, 9, 12, 15, 20, 24, 36, 48, 60, 90, 110, 127, 150, 220, 240, 300, 380, 400, 450, 500]

omega = np.linspace(100, 15000, 500)
all_combinations = list(itertools.product(R_list, L_list, C_list, U_list))
selected_combinations = random.sample(all_combinations, 100)
```

```python
for i, (R, L, C, U_in) in enumerate(selected_combinations):
    X_L = omega * L
    X_C = 1 / (omega * C)
    Z = np.sqrt(R**2 + (X_L - X_C)**2)
    I = U_in / Z
    
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(omega, I, color='blue', linewidth=2)
    
    C_mkF = C * 1e6
    title = f"R={R} Ом, L={L} Гн, C={C_mkF:.1f} мкФ, U={U_in} В"
    ax.set_title(title, fontsize=10)
    ax.set_xlabel('Угловая частота ω (рад/с)')
    ax.set_ylabel('Ток (А)')
    ax.grid(True)
    
    filename = f"{output_dir}/graph_{i+1}_R{R}_L{L}_C{C_mkF:.1f}_U{U_in}.png"
    plt.savefig(filename, format='png', bbox_inches='tight')
    plt.close(fig)
```
