```python
import pandas as pd
import numpy as np
import itertools
import time
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python

R_list = [1, 2, 4, 8, 10, 15, 20, 25, 30, 40, 50, 60, 75, 100, 120, 150, 200, 300, 400, 500]
L_list = [0.005, 0.01, 0.02, 0.03, 0.05, 0.07, 0.1, 0.12, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0]
C_list = [1e-7, 2e-7, 5e-7, 1e-6, 2e-6, 3e-6, 4e-6, 5e-6, 8e-6, 1e-5, 1.5e-5, 2e-5, 3e-5, 4e-5, 5e-5, 6e-5, 7e-5, 8e-5, 9e-5, 1e-4]
U_list = [5, 9, 12, 15, 20, 24, 36, 48, 60, 90, 110, 127, 150, 220, 240, 300, 380, 400, 450, 500]

omega = np.linspace(100, 15000, 500)
data = []
start_time = time.time()

total = len(R_list) * len(L_list) * len(C_list) * len(U_list)
```

```python
print(f"Generation begins")

for R, L, C, U_in in itertools.product(R_list, L_list, C_list, U_list):
    X_L = omega * L
    X_C = 1 / (omega * C)
    Z = np.sqrt(R**2 + (X_L - X_C)**2)
    I = U_in / Z
    
    I_max = np.max(I)
    omega_0 = 1 / np.sqrt(L * C)
    
    data.append({
        'R': R,
        'L': L,
        'C': C,
        'U': U_in,
        'I_max': I_max,
        'omega_0': omega_0,
        'Z_min': R
    })

df = pd.DataFrame(data)

noise_normal = np.random.normal(loc=0, scale=0.2, size=len(df))

df['I_max'] = df['I_max'] + noise_normal
outliers_idx = df.sample(frac=0.001).index
df.loc[outliers_idx, 'I_max'] = df.loc[outliers_idx, 'I_max'] * np.random.uniform(10, 50, size=len(outliers_idx))

df.to_csv('rlc_dataset_noisy.csv',index=False)


end_time = time.time()
print(f"Datased created by {end_time - start_time:.2f} seconds")
print(f"Size of dataset: {df.shape[0]} rows, {df.shape[1]} cols.\n")
```
